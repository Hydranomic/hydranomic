# Hydranomic - Processador de Gráficos de Consumo Anual

Notebook para consolidação e formatação dos dados de consumo mensal de água em formato de payload JSON para a plataforma **WEGnology**.
Gera o arquivo `grafico_anual.json` estruturado com os 12 meses do ano.

```python
import os
import sys
import json
from pathlib import Path
import matplotlib.pyplot as plt

from data_chart import processar_dados_grafico
from wegnology_adapter import carregar_json, salvar_json
```

## 1. Processamento e Consolidação dos Dados

```python
# Executa o processamento lendo automaticamente de INPUT_DIR
resultado_grafico = processar_dados_grafico(salvar=True)

print(f"Ano Processado: {resultado_grafico['ano_referencia']}")
print(f"Mês Atual: {resultado_grafico['mes_atual']}")
print(f"Unidade de Medida: {resultado_grafico['unidade_medida']}")

# Exibe a tabela dos 12 meses
for m in resultado_grafico["meses"]:
    val = f"{m['consumo_m3']} m³" if m['consumo_m3'] is not None else "Sem leitura"
    print(f" - {m['rotulo']} (Mês {m['mes']:02d}): {val}")
```

## 2. Visualização do Gráfico Anual

```python
rotulos = [m['rotulo'] for m in resultado_grafico['meses']]
valores = [m['consumo_m3'] if m['consumo_m3'] is not None else 0 for m in resultado_grafico['meses']]
cores = ['#0284c7' if m['consumo_m3'] is not None else '#e2e8f0' for m in resultado_grafico['meses']]

plt.figure(figsize=(11, 4.5))
barras = plt.bar(rotulos, valores, color=cores, edgecolor='#0369a1')
plt.title(f"Consumo Mensal de Água - Ano {resultado_grafico['ano_referencia']}", fontsize=14)
plt.xlabel("Mês")
plt.ylabel("Consumo (m³)")

# Adiciona valores em cima das barras preenchidas
for barra, val in zip(barras, valores):
    if val > 0:
        plt.text(barra.get_x() + barra.get_width()/2., val + 5, f"{int(val)}", ha='center', va='bottom', fontsize=9)

plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```
